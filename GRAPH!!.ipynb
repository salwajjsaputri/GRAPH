1.	Representasi Graf dengan Adjacency List

```python
# Membuat graf tak berarah menggunakan adjacency listgraph =  {
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

# menampikan graf
def print_graph(graf):
    for node in graf:
        print(f"{node} -> {graf[node]}")

print_graph(graph)
```

2.	Traversal Graf menggunakan BFS

```python
from collections import deque

def bfs(graph, start):
    visited = set()
    queue = deque([start])

    while queue:
        vertex = queue.popleft()
        if vertex not in visited:
            print(vertex, end=" ")
            visited.add(vertex)
            queue.extend([neighbor for neighbor in graph[vertex] if neighbor not in visited])

# contoh pemanggilan
bfs(graph, 'A')
```

3.	Traversal Graf menggunakan DFS

```python
def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()

    if start not in visited:
        print(start, end=" ")
        visited.add(start)
        for neighbor in graph[start]:
            dfs(graph, neighbor, visited)

# contoh pemanggilan
dfs(graph, 'A')
```

```python
from collections import deque

# BFS
def bfs(graph, start):
    visited = set()
    queue = deque([start])
    while queue:
        v = queue.popleft()
        if v not in visited:
            print(v, end=" ")
            visited.add(v)
            for n in graph[v]:
                if n not in visited:
                    queue.append(n)
    print()

# DFS
def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()
    if start not in visited:
        print(start, end=" ")
        visited.add(start)
        for n in graph[start]:
            dfs(graph, n, visited)

# jalanakan
print("BFS:", end=" ")
bfs(graph, 'A')

print("DFS:", end=" ")
dfs(graph, 'A')
print()
```

```python
def bfs_return_list(graph, start):
    visited = set()
    queue = deque([start])
    order = []
    while queue:
        vertex = queue.popleft()
        if vertex not in visited:
            order.append(vertex)
            visited.add(vertex)
            queue.extend([n for n in graph[vertex] if n not in visited])
    return order

print("Urutan BFS:", bfs_return_list(graph, 'A'))
```

```python
def find_path(graph, start, end, path=None):
    if path is None:
        path = []
    path = path + [start]
    if start == end:
        return path
    if start not in graph:
        return None
    for neighbor in graph[start]:
        if neighbor not in path:
            new_path = find_path(graph, neighbor, end, path)
            if new_path:
                return new_path
    return None

print("Jalur A ke E:", find_path(graph, 'A', 'E'))
print("Jalur A ke D:", find_path(graph, 'A', 'D'))
print("Jalur B ke C:", find_path(graph, 'B', 'C'))
```

```python
def is_connected(graph):
    if not graph:
        return True
    start = next(iter(graph))
    visited = set()

    def dfs_check(node):
        visited.add(node)
        for neighbor in graph[node]:
            if neighbor not in visited:
                dfs_check(neighbor)

    dfs_check(start)
    return len(visited) == len(graph)

print("Apakah graf terhubung?", is_connected(graph))
```
